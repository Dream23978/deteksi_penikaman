Import Model Yolo11s

```python
from ultralytics import YOLO

model = YOLO('yolo11s.pt')
```

Training Model

```python

results = model.train(
    data='jun_knife_weapon-20/data.yaml',
    epochs=35,
    imgsz=640,
    batch=4,
    device=0,
    workers=1,
    plots=True,
    erasing=0.4,
    mosaic=0.5,
    mixup=0.1,
    scale=0.5,
)
```

Evaluasi Model

```python
# Evaluasi performa model akhir
metrics = model.val(data='jun_knife_weapon-20/data.yaml', conf=0.35)

print("\n" + "="*45)
print("     Classification Report     ")
print("="*45)
print(f"Akurasi Deteksi (mAP50)    : {metrics.results_dict['metrics/mAP50(B)']:.4f}")
print(f"Tingkat Presisi (Precision): {metrics.results_dict['metrics/precision(B)']:.4f}")
print(f"Daya Tangkap Objek (Recall) : {metrics.results_dict['metrics/recall(B)']:.4f}")
print("="*45)

from IPython.display import Image, display
import os
# Menampilkan grafik Confusion Matrix dari folder run terakhir
save_dir = model.trainer.save_dir if hasattr(model, 'trainer') and model.trainer else 'runs/detect/train'
cm_path = os.path.join(save_dir, 'confusion_matrix.png')
if os.path.exists(cm_path):
    print(f"\n[INFO] Menampilkan Confusion Matrix dari {cm_path}:")
    display(Image(filename=cm_path))
```
